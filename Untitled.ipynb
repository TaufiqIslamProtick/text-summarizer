```python
from rouge_score import rouge_scorer

# Define the reference and summary
reference = """
Works very well to track steps, calories, distance and work outs. Data is accurate and battery life is excellent. 
I wear it 24 hours and the sleep data has been very helpful. I also use the smart wake feature so an alarm doesn't 
wake my partner since I get up a couple hours before him. The smart wake sends vibrations which buzz the arm. 
It works surprisingly well. Even though there is a water lock to allow user to wear in shower or swimming, I have 
not tried it yet, but plan to. The sensors on the sides are easy to use for the EDA scan and heart rhythm scan. 
I would prefer to have the apps in a different order but that is just a personal thing. The only other issue I have 
had is the phone app occasionally acts up and won't allow me to sync data. It self resolves but can be annoying. 
I chalk it up to consequence of having so much tech these days. I am glad I chose the 5 over the 4 version for 
the built in GPS & water lock capabilities. I also recommend getting the fitbit premium account if you are serious 
about getting your health better. It has additional insights and data analyses that you won't have access to 
without it. Overall extremely happy with the Fitbit 5 and I do recommend it.
"""

summary = """
The user is highly satisfied with the Fitbit 5, commending its accurate tracking of steps, calories, distance, and 
workouts, as well as its excellent battery life. They benefit from wearing it 24 hours daily, utilizing features 
like sleep tracking and the smart wake alarm, which uses vibrations to avoid disturbing others. Although they 
haven't yet used the water lock feature, they appreciate the easy-to-use sensors for EDA and heart rhythm scans. 
Minor issues include occasional syncing problems with the phone app and a preference for a different app layout. 
They prefer the Fitbit 5 over the 4 version for its built-in GPS and water lock capabilities and recommend 
subscribing to Fitbit Premium for additional health insights. Overall, they strongly recommend the Fitbit 5.
"""

# Initialize the ROUGE scorer
scorer = rouge_scorer.RougeScorer(['rouge1', 'rouge2', 'rougeL'], use_stemmer=True)

# Calculate ROUGE scores
scores = scorer.score(reference, summary)
print(scores)
```

```python
import bert_score

# Hypotheses (model-generated summaries)
summaries = [
    "The Fitbit 5 is great for tracking steps, calories, and workouts.",
    "It has excellent battery life and useful features."
]

# References (original texts)
reviews = [
    "Fitbit 5 works well for tracking steps, calories, distance, and workouts. The battery life is excellent.",
    "It has features like sleep tracking and smart wake. The battery lasts long."
]

# Calculate BERTScore
P, R, F1 = bert_score.score(summaries, reviews, lang="en", rescale_with_baseline=True)

# Print the F1 scores (most common metric for BERTScore)
for i, f1_score in enumerate(F1):
    print(f"Summary {i+1} BERTScore F1: {f1_score:.4f}")
```
